# Estimasi Tambahan Pembelian Akibat Iklan

## Metodologi

1. Memfilter data untuk masing-masing grup (ad dan psa).
2. Menghitung jumlah konversi dan jumlah pengguna masing-masing grup.
3. Menghitung conversion rate masing-masing grup.
4. Menghitung estimasi tambahan pembelian akibat iklan.

## Persiapan Data dan Library

Mengimpor library yang dibutuhkan dan memuat dataset untuk proses analisis.

```python
# Mengimpor library
import pandas as pd
import warnings
warnings.filterwarnings(action='ignore', category=FutureWarning)

# Memuat data
df = pd.read_csv('Data/marketing_AB.csv')
df.drop(columns='Unnamed: 0', inplace=True)
df.columns = df.columns.str.replace(' ', '_')
```

## Filter Data Masing-Masing Grup

Memfilter data untuk grup ad (`df_ad`) dan grup psa (`df_psa`) untuk memudahkan perhitungan.

```python
df_ad = df[df['test_group'] == 'ad'].copy()
df_psa = df[df['test_group'] == 'psa'].copy()
```

## Hitung Jumlah Konversi dan Jumlah Pengguna

Menghitung jumlah konversi (`X_ad`, `X_psa`) dan jumlah pengguna (`n_ad`, `n_psa`) masing-masing grup.

```python
# Grup ad
X_ad, n_ad = df_ad['converted'].sum(), len(df_ad)

# Grup psa
X_psa, n_psa = df_psa['converted'].sum(), len(df_psa)
```

## Hitung Conversion Rate

Menghitung conversion rate grup ad (`p_ad`) dan grup psa (`p_psa`).

```python
# Menghitung conversion rate masing-masing grup
p_ad = X_ad / n_ad
p_psa = X_psa / n_psa

print(f'Conversion rate grup ad: {p_ad * 100:.2f}%')
print(f'Conversion rate grup psa: {p_psa * 100:.2f}%')
print(f'Selisih conversion rate: {(p_ad - p_psa) * 100:.2f}%')
```

## Estimasi Tambahan Pembelian

Menghitung estimasi tambahan pembelian akibat iklan.

```python
additional_purchase_estimated = (p_ad - p_psa) * n_ad

print(f'Estimasi tambahan pembelian: {additional_purchase_estimated:,.0f} pembelian')
```
