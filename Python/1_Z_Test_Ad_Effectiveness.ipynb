# Uji Statistik Efektivitas Iklan terhadap Conversion Rate

## Metodologi

1. Memfilter data untuk masing-masing grup (ad dan psa).
2. Menghitung jumlah konversi dan jumlah pengguna masing-masing grup.
3. Melakukan Z-test.

## Persiapan Data dan Library

Mengimpor library yang dibutuhkan dan memuat dataset untuk proses analisis.

```python
# Mengimpor library
import pandas as pd
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
import warnings
warnings.filterwarnings(action='ignore', category=FutureWarning)

# Memuat data
df = pd.read_csv('Data/marketing_AB.csv')
df.drop(columns='Unnamed: 0', inplace=True)
df.columns = df.columns.str.replace(' ', '_')
```

## Filter Data Masing-Masing Grup

Memfilter data grup ad (`df_ad`) dan grup psa (`df_psa`) untuk memudahkan perhitungan.

```python
df_ad = df[df['test_group'] == 'ad'].copy()
df_psa = df[df['test_group'] == 'psa'].copy()
```

## Hitung Jumlah Konversi dan Jumlah Pengguna

Menghitung jumlah konversi (`X_ad`, `X_psa`) dan jumlah pengguna (`n_ad`, `n_psa`) masing-masing grup.

```python
# Grup ad
X_ad, n_ad = df_ad['converted'].sum(), len(df_ad)

# Grup psa
X_psa, n_psa = df_psa['converted'].sum(), len(df_psa)
```

## Z-Test

Melakukan Z-test untuk mendapatkan Z-score dan p-value.

```python
# Jumlah konversi
counts = np.array([X_ad, X_psa])

# Jumlah pengguna
nobs = np.array([n_ad, n_psa])

# Z-test
z_score, p_value = proportions_ztest(counts, nobs, alternative='larger')

print(f'Z-score: {z_score}')
print(f'p-value: {p_value}')
```
